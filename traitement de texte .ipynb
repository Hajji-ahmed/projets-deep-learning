```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.utils import pad_sequences
# Téléchargement d'un dataset textuel
path_to_file = tf.keras.utils.get_file('shakespeare.txt', 'https://storage.googleapis.com/download.tensorflow.org/data/shakespeare.txt')
# Chargement des données
with open(path_to_file, 'r') as file:
 text_data = file.read()
# Aperçu des données
print(f"Longueur totale du texte : {len(text_data)} caractères")
print(f"Extrait du texte : \n{text_data[:500]}")
```

```python
# Tokenisation des mots(unique)
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text_data])
# Conversion des mots en séquences d'indices
sequences = tokenizer.texts_to_sequences([text_data])[0]
# Taille du vocabulaire
vocab_size = len(tokenizer.word_index) + 1
print(f"Nombre total de mots uniques (vocabulaire) : {vocab_size}")
# Création des séquences d'entrée et de cible
sequence_length = 5
input_sequences = []
for i in range(sequence_length, len(sequences)):
 input_sequences.append(sequences[i-sequence_length:i+1])
# Conversion en tenseurs
input_sequences = tf.constant(input_sequences)
# Séparation des entrées et des cibles
X = input_sequences[:, :-1]
y = input_sequences[:, -1]
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
# Construction du modèle
3
model = Sequential([
 Embedding(input_dim=vocab_size, output_dim=64, 
input_length=sequence_length), # Couche embedding
 SimpleRNN(128, return_sequences=False), # Couche RNN
 Dense(vocab_size, activation='softmax') # Couche de sortie
])
# Compilation
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', 
metrics=['accuracy'])
# summary
model.summary()
```

```python
history = model.fit(X, y, epochs=10, batch_size=64, validation_split=0.2)
```

```python
# Génération de texte exemple
seed_text = "Il était une fois"
num_words_to_generate = 50

for i in range(num_words_to_generate):
    # Convertir le texte en séquence
    token_list = tokenizer.texts_to_sequences([seed_text])[0]

    # Compléter avec un padding si nécessaire
    token_list = pad_sequences([token_list], maxlen=sequence_length-1, padding='pre')

    # Prédire le mot suivant
    predicted = model.predict(token_list, verbose=0)
    predicted_word_index = tf.argmax(predicted, axis=-1).numpy()[0]

    # Obtenir le mot correspondant
    output_word = ""
    for word, index in tokenizer.word_index.items():
        if index == predicted_word_index:
            output_word = word
            break

    # Ajouter le mot prédit au texte
    seed_text += " " + output_word

print(seed_text)
```

```python
def generate_text_prediction(model, tokenizer, seed_text, sequence_length, num_words_to_generate):
    for _ in range(num_words_to_generate):
        # Convertir le texte en séquence
        token_list = tokenizer.texts_to_sequences([seed_text])[0]
        # Compléter avec un padding si nécessaire
        token_list = pad_sequences([token_list], maxlen=sequence_length-1, padding='pre')
        # Prédire le mot suivant
        predicted = model.predict(token_list, verbose=0)

        predicted_word_index = tf.argmax(predicted, axis=-1).numpy()[0]
        #predicted_word_index = 1  # Car 0.3 est la probabilité maximale

        # Obtenir le mot correspondant
        output_word = ""
        for word, index in tokenizer.word_index.items():
            if index == predicted_word_index:
                output_word = word
                break
        # Ajouter le mot prédit au texte
        seed_text += " " + output_word
    return seed_text

# Exemple d'utilisation
generated_text = generate_text_prediction(model, tokenizer, "Il était une fois", sequence_length, 50)
print(generated_text)
```
